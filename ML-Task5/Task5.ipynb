```python
import pandas as pd
```

```python
data = {
    "Name": ["Ali", "ALI", "sara", "Ahmed", "Sara", None,
             "Usman", "Hina", "Bilal", "Ali", "Zoya", "Omar"],

    "Age": ["20", "20", "21", "22", "21", "19",
            "150", "20", "twenty", "20", "23", "24"],

    "Marks": ["85", "85", "90", "78", "90", "88",
              "75", "500", "82", None, "91", "76"],

    "City": ["lahore", "lahore", "KARACHI", "islamabad",
             "karachi", "LAHORE", "multan", "Multan",
             "LAHORE", "lahore", "faisalabad", "ISLAMABAD"]
}

df = pd.DataFrame(data)

df.to_csv("messy_students.csv", index=False)

print(df)
```

```python
df = pd.read_csv("messy_students.csv")

print(df.head())
```

```python
print("Missing values:")
print(df.isnull().sum())

# Fill missing names
df["Name"] = df["Name"].fillna("Unknown")

# Convert age and marks into numbers
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# Fill missing values with median
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())

print("Missing values fixed!")
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows found:", duplicates)

df = df.drop_duplicates()

print("Duplicates removed:", duplicates)
```

```python
df["Name"] = df["Name"].str.strip().str.title()

df["City"] = df["City"].str.strip().str.title()

print(df[["Name", "City"]])
```

```python
print(df.dtypes)
```

```python
for col in ["Age", "Marks"]:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    df[col] = df[col].clip(lower, upper)

    print(col, "outliers fixed:", outliers)
```

```python
print("Cleaned Data:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv("clean_students.csv", index=False)

print("Clean file saved successfully!")
```

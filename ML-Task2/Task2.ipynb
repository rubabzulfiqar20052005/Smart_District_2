```python
import pandas as pd
```

```python
data = {
    "product": ["Laptop", "Mobile", "Mouse", "Laptop", "Keyboard",
                "Mobile", "Mouse", "Laptop", "Keyboard", "Mobile"],
    "region": ["North", "South", "North", "South", "North",
               "North", "South", "North", "South", "North"],
    "date": ["2026-01-01", "2026-01-02", "2026-01-03", "2026-01-04",
             "2026-01-05", "2026-01-06", "2026-01-07", "2026-01-08",
             "2026-01-09", "2026-01-10"],
    "quantity": [2, 5, 10, 3, 4, 6, 8, 1, 7, 3],
    "price": [1000, 500, 20, 1000, 50, 500, 20, 1000, 50, 500]
}

df = pd.DataFrame(data)

df.to_csv("sales.csv", index=False)

print("Sales CSV file created successfully!")
print(df)
```

```python
df = pd.read_csv("sales.csv")

print("First 5 Rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

```python
north_data = df[df["region"] == "North"]

print("Sales Data for North Region:")
print(north_data)

print("\nShape of North Data:")
print(north_data.shape)
```

```python
north_data = north_data.copy()

north_data["revenue"] = north_data["quantity"] * north_data["price"]

print("Sales Data with Revenue:")
print(north_data)

print("\nShape:", north_data.shape)
```

```python
product_sales = north_data.groupby("product")["revenue"].sum()

print("Total Revenue by Product:")
print(product_sales)
```

```python
top_products = product_sales.sort_values(ascending=False).head(5)

print("Top 5 Products by Revenue:")
print(top_products)
```

```python
metadata = pd.DataFrame({
    "product": ["Laptop", "Mobile", "Mouse", "Keyboard"],
    "category": ["Electronics", "Electronics", "Accessories", "Accessories"],
    "brand": ["Dell", "Samsung", "Logitech", "HP"]
})

print("Product Metadata:")
print(metadata)
```

```python
final_data = pd.merge(
    north_data,
    metadata,
    on="product",
    how="left"
)

print("Final Merged Data:")
print(final_data)

print("\nFinal Data Shape:")
print(final_data.shape)
```

```python
final_data.to_csv("final_sales.csv", index=False)

print("Final sales data saved to final_sales.csv")
```
